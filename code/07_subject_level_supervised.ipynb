# Step 7 - Supervised subject-level classification (paper Sec. 3.5.4, Tables 3, 8, 9)

* Splits the training subjects into two parts (`train1`, `train2`): `train1` is used to train the
  generator and the utterance classifier, `train2` to train the subject-level classifiers.
* For every subject, counts how many of its utterances the utterance classifier assigns to each class.
  These four counts are the features for the subject-level classifier.
* Subject-level classifiers (scikit-learn): Gaussian NB, Multinomial NB, Decision Tree,
  Random Forest, KNN, SVM. Evaluated on the 120 test subjects.
* The first half of the notebook uses the CNN trained on real data only (`CNN_pw_half.h5`);
  the second half uses CNNs trained on real + generated data (`CNN_bal2000_half.h5`, `CNN_half_bal1000.h5`).

*Original file name: `Untitled5.ipynb`. Code cells are unchanged from 2019/2020; outputs were cleared and these markdown notes were added in 2026.*

## Subject features from utterance predictions (CNN trained on real data)

```python
import tensorflow as tf
from keras.models import model_from_json
import numpy as np
import operator

CNN_class = tf.keras.models.load_model("CNN_pw_half.h5")

vectors_file = "train2_vectors.npy"
labels_file = "train2_labels.npy"
subjId_file = "train2_subj.npy"

vectors = np.load(vectors_file)
labels = np.load(labels_file)
subjId = np.load(subjId_file)

subjDict = dict()
subjIdDict = dict()
subjLabel = dict()

for n in range(len(labels)):
    vec = np.array([vectors[n]])
    lab = labels[n]
    sub = subjId[n]
    prediction = CNN_class.predict(vec)
    pred_lab = prediction.argmax(axis=-1)[0]
    if sub in subjDict:
        subjDict[sub].append(pred_lab)
    else:
        subjDict[sub] = [pred_lab]
        subjLabel[sub] = lab

correct = 0
incorrect = 0

feature_dict = dict()

for subj in subjDict:
    d = dict()
    preds = subjDict[subj]
    for p in preds:
        if p in d:
            d[p] += 1
        else:
            d[p] = 1
    if 3 in d:
        d[3] = d[3]
    feature_dict[subj] = d
    sorted_d = sorted(d.items(), key=operator.itemgetter(1))
    prediction = sorted_d[::-1][0][0]
    gt = subjLabel[subj]
    if prediction == gt:
        correct += 1
    else:
        incorrect += 1

print(correct, incorrect, correct/(correct+incorrect))


```
