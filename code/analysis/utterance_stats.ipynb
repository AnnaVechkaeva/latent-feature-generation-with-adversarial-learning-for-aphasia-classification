# Dataset statistics: number of utterances and average utterance length per class

Reads the cleaned transcripts. Not needed to reproduce the results.

*Code cells are unchanged from 2019/2020; outputs were cleared and these markdown notes were added in 2026.*

```python
import os, nltk
```

```python
aph_paths = []
aph_ids = []

for root, dirs, files in os.walk("./clean/", topdown=False):
    for name in files:
        if name != ".DS_Store" and name != "0metadata.cdc":
            aph_paths.append(''.join([root, name]))
            aph_ids.append(name[:-6])
```

```python
texts = []
for r in aph_paths:
    with open(r) as file:
        lines = file.readlines()
        texts.append(lines)
```

```python
label_dict = dict()

with open("id_label_aph.txt") as file:
    id_label = file.read()
    id_label = id_label.split('\n')
    id_label = id_label[:-1]
    for i in id_label:
        _id = i.split('\t')[0]
        label = i.split('\t')[1]
        label_dict[_id] = label
```

```python
utt_n = dict()
for t in range(len(texts)):
    _id = aph_ids[t]
    text = texts[t]
    label= label_dict[_id]
    if label in utt_n:
        utt_n[label] += len(text)
    else:
        utt_n[label] = len(text)
print(utt_n)
```

```python
lab_utt = dict()

for n in range(len(texts)):
    text = texts[n]
    _id = aph_ids[n]
    label = label_dict[_id]
    if label in lab_utt:
        lab_utt[label] += text
    else:
        lab_utt[label] = text
```

```python
lab_lenw = dict()

for l in lab_utt:
    utts = lab_utt[l]
    utt_w = [nltk.word_tokenize(i) for i in utts]
    utt_lenw = [len(i) for i in utt_w]
    utt_lenc = [len(i) for i in utts]
    print(l)
    print(sum(utt_lenw)/len(utt_lenw))
    print(sum(utt_lenc)/len(utt_lenc))
```
