# Dataset statistics: average WAB score and age per aphasia type / controls

Reads the raw AphasiaBank transcripts. Not needed to reproduce the results.

*Original file name: `Dataset_stats.ipynb`. Code cells are unchanged from 2019/2020; outputs were cleared and these markdown notes were added in 2026.*

```python
import os, re
```

```python
aph_paths = []
aph_ids = []

for root, dirs, files in os.walk("./aphasia/English_scripts/", topdown=False):
    for name in files:
        if name != ".DS_Store" and name != "0metadata.cdc":
            aph_paths.append('/'.join([root, name]))
            aph_ids.append(name[:-4])
```

```python
label_dict = dict()

with open("id_label_aph.txt") as file:
    id_label = file.read()
    id_label = id_label.split('\n')
    id_label = id_label[:-1]
    for i in id_label:
        _id = i.split('\t')[0]
        label = i.split('\t')[1]
        label_dict[_id] = label
```

```python
wab_dict = dict()
age_dict = dict()

for n in range(len(aph_paths)):
    with open(aph_paths[n]) as file:
        print(aph_ids[n], label_dict[aph_ids[n]])
        label = label_dict[aph_ids[n]]
        lines = file.readlines()
        par_info = ''
        for i in lines:
            if "@ID:" in i and "PAR" in i:
                par_info = i
                break
        print(par_info)
        wab = re.search("\|([0-9]{2}\.[0-9])\|$", par_info)
        age = re.search("PAR\|([0-9]{2})\;[0-9]{2}\.\|", par_info)
        if wab != None:
            wab = float(wab.group(1))
            print(wab)
            if label in wab_dict:
                wab_dict[label].append(wab)
            else:
                wab_dict[label] = [wab]
        if age != None:
            age = float(age.group(1))
            print(age)
            if label in age_dict:
                age_dict[label].append(age)
            else:
                age_dict[label] = [age]
        print('')
```

```python
for i in wab_dict:
    print(i)
    wab_list = wab_dict[i]
    avg_wab = sum(wab_list)/len(wab_list)
    print(avg_wab)
    print('')
```

```python
for i in age_dict:
    print(i)
    age_list = age_dict[i]
    avg_age = sum(age_list)/len(age_list)
    print(avg_age)
    print('')
```

```python
cont_paths = []
cont_ids = []

for root, dirs, files in os.walk("./Control/", topdown=False):
    for name in files:
        if name != ".DS_Store" and name != "0metadata.cdc":
            cont_paths.append('/'.join([root, name]))
            cont_ids.append(name[:-4])
```

```python
age_cont = []

for n in range(len(cont_paths)):
    with open(cont_paths[n]) as file:
        print(cont_paths[n]) #label_dict[cont_ids[n]])
        #label = label_dict[cont_ids[n]]
        lines = file.readlines()
        par_info = ''
        for i in lines:
            if "@ID:" in i and "PAR" in i:
                par_info = i
                break
        print(par_info)
        age = re.search("PAR\|([0-9]{2})\;[0-9]{2}\.\|", par_info)
        if age != None:
            age = float(age.group(1))
            print(age)
            age_cont.append(age)
        print('')
```
