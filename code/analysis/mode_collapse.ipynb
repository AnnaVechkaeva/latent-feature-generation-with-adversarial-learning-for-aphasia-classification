# Checking generated data for mode collapse (paper Sec. 5)

PCA and t-SNE of averaged utterance vectors: the 10k-per-class set (`_10k`), plotted class by class, and the real training data.

*Original file name: `mode_collapse.ipynb`. Code cells are unchanged from 2019/2020; outputs were cleared and these markdown notes were added in 2026.*

```python
import numpy as np
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
%matplotlib inline  
```

```python
vectors = np.load("aphasia_np_vectors_patientwise_train_10k.np.npy")
labels = np.load("aphasia_np_labels_patientwise_train_10k.np.npy")
```

```python
avg_vectors = [np.mean(i, axis=0) for i in vectors]
principalComponents = pca.fit_transform(avg_vectors)
```

```python
pca1 = [i[0] for i in principalComponents]
pca2 = [i[1] for i in principalComponents]
```

```python
plt.scatter(pca1[:10000], pca2[:10000], color='darkgreen')
plt.show()
```

```python
plt.scatter(pca1[10000:20000], pca2[10000:20000], color='darkgreen')
plt.show()
```

```python
plt.scatter(pca1[20000:30000], pca2[20000:30000], color='darkgreen')
plt.show()
```

```python
plt.scatter(pca1[30000:40000], pca2[30000:40000], color='darkgreen')
plt.show()
```

```python
from sklearn.manifold import TSNE
```

```python
X_embedded = TSNE(n_components=2).fit_transform(avg_vectors)
```

```python
tsne1 = [i[0] for i in X_embedded]
tsne2 = [i[1] for i in X_embedded]
```
